```sql
-- ============================================================
-- Archivo: 00_setup_environment.sql
-- Propósito:
-- Crear la estructura base de Unity Catalog requerida
-- por el proyecto de eventos de ecommerce.
--
-- Ejecutar una única vez por ambiente.
-- ============================================================


-- ------------------------------------------------------------
-- 1. Crear el catálogo
-- ------------------------------------------------------------
CREATE CATALOG IF NOT EXISTS ecommerce_history
COMMENT 'Catálogo del proyecto de análisis histórico de ecommerce';


-- ------------------------------------------------------------
-- 2. Seleccionar el catálogo
-- ------------------------------------------------------------
USE CATALOG ecommerce_history;


-- ------------------------------------------------------------
-- 3. Crear el esquema
-- ------------------------------------------------------------
CREATE SCHEMA IF NOT EXISTS events_history
COMMENT 'Esquema para las tablas y archivos históricos de eventos';


-- ------------------------------------------------------------
-- 4. Seleccionar el esquema
-- ------------------------------------------------------------
USE SCHEMA events_history;


-- ------------------------------------------------------------
-- 5. Crear el volumen administrado
-- ------------------------------------------------------------
CREATE VOLUME IF NOT EXISTS db_ecommerce_history
COMMENT 'Volumen para almacenar los archivos CSV de eventos de ecommerce';


-- ------------------------------------------------------------
-- 6. Validar el ambiente
-- ------------------------------------------------------------
SELECT
    current_catalog() AS catalogo_actual,
    current_schema() AS esquema_actual;
```
